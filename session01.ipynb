1. How many passanger is under 22 years old?

```python
train_df["Age"].isnull().value_counts()
```

There are about 177 null values in the Age column. Missing ages count as `False` in the comparision, so they're left out

```python
not_adults = (train_df["Age"] < 22.0).sum()
# use .sum() not .count() because .count() returns the total number of rows in
# the column and doesn't look at the condition
print("Passanger that are not an adult:", not_adults)
```

2. Did woman survive more than men?

```python
survivor = train_df.groupby("Sex")["Survived"].mean()
# to displayed the value of both sex and survived, use value_counts()
# if we use count() it will only displayed the value of the first column and
# the total of the second
# raw counts can mislead. to compare groups of different sizes, we need rate,
# use .mean() (shorter way of doing normalize=true to find rate). survived can
# use the shorter way because `Survived` is 1s and 0s, so its mean is the survival rate
survivor
```

3. Survival rate by sex and class together

```python
survival_rate = train_df.groupby(['Sex', 'Pclass'])['Survived'].value_counts(normalize=True)
# use normalize=True to get percentage
survival_rate
```

4. Did people who paid more survived more?

```python
survived_fare = train_df.groupby(["Fare"])["Survived"].mean()
survived_fare.sort_values(ascending=False).head()
```

5. Did more people travel alone or with families?

```python
total_pass = train_df["Name"].count()
print("Total passangers on board:", total_pass)

train_df["alone"] = train_df["SibSp"] + train_df["Parch"]
alone = (train_df["alone"] == 0).sum()
print("Total passangers on baord that come alone:", alone)
```
